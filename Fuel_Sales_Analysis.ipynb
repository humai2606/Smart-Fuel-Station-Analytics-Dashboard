```python
import pandas as pd
import matplotlib.pyplot as plt
fuel = ["Petrol", "Diesel", "LPG"]
revenue = [85000, 62000, 45000]
quantity = [900, 650, 300]
df = pd.DataFrame({
    "Fuel Type": fuel,
    "Revenue": revenue,
    "Quantity Sold": quantity
})
print(df)
plt.figure(figsize=(6,5))
plt.bar(df["Fuel Type"], df["Revenue"])
plt.xlabel("Fuel Type")
plt.ylabel("Revenue")
plt.title("Fuel Sales Revenue Analysis")
for i in range(len(fuel)):
    plt.text(i, revenue[i], revenue[i])
plt.show()
print("\nInsights:")
print("1. Petrol generates highest revenue.")
print("2. Diesel sales are moderate.")
print("3. LPG has lower quantity but higher unit price.")
```
